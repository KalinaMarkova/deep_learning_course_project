# Fine-Grained Analysis of Propaganda in News Articles
## Notebook 07: Recap of Experiment Results

Here is a recap of the results of all experiments as per the evaluation metrics of the  SemEval-2020 Task 11.

### Summary of Propaganda Detection Experiments



| Experiment | Architecture | Gold Spans | Predicted Spans | Precision | Recall | F1 Score |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **Exp 1** | **Joint Tagger** | 587 | 2,505 | 16.21% | **44.36%** | 23.74% |
| **Exp 2** | **Two-Stage Pipeline** | 587 | 1,021 | 18.67% | 29.79% | 22.95% |
| **Exp 3** | **Multi-Task Learning** | 662 | 404 | 26.39% | 21.57% | 23.74% |
| **Exp 4** | **Two-Stage + CRF (Gap=4)** | 563 | 465 | **47.00%** | 25.98% | **33.46%** |

---

#### Some observations:

*   **Exp 1:** The Joint Tagger achieved massive recall by aggressively predicting 2,505 spans. However, this flooded the evaluation with false positives and fragmented boundaries, driving precision down to 16.21%.
*   **Exp 2:** Moving to a Two-Stage Pipeline reduced the number of false positives  (predicting 1,021 spans), but it failed to significantly improve precision due to cascading errors. If the first model passed a bad boundary, the second model was guaranteed to fail.
*   **Exp 3:** Multi-Task Learning forced the model to share representations, making it highly conservative. It predicted only 404 spans, which boosted precision to 26.39% but reduced recall.
*   **Exp 4:** The introduction of Conditional Random Fields (CRF) and boundary smoothing (merging spans separated by 4 or fewer tokens) gave the best result. By stitching fragmented predictions together, precision jumped to 47.00%, securing the highest overall F1 score of 33.46%.

Such sequence tagging tasks, such as propaganda spans are **very subjective** even for humans. Raw neural architectures struggle with boundary fragmentation. A two-stage pipeline augmented with intelligent boundary smoothing (CRF and gap-merging) yielded the most robust performance, effectively doubling precision without severely sacrificing core recall.
